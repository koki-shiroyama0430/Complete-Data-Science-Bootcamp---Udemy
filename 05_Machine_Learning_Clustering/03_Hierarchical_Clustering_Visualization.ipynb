```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
data = pd.read_csv('Country clusters standardized.csv',index_col = 'Country')
```

```python
x_scaled = data.copy()
x_scaled = x_scaled.drop(['Language'],axis =1)
x_scaled
```

```python
sns.clustermap(x_scaled,cmap='mako')
```
