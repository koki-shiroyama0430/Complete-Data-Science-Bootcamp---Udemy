Import libraries

```python
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
#Generate some data we are going to split
a = np.arange(1,101)
```

```python
a
```

```python
b = np.arange(501,601)
b
```

```python
#Split the data
train_test_split(a)
```

```python
a_train, a_test = train_test_split(a)
```

```python
a_train.shape,a_test.shape
```

```python
a_train
```

```python
a_test
```

```python
# デフォルトで75：25になるが、テストデータが多いので、宣言時に割合を指定する
a_train, a_test = train_test_split(a,test_size=0.2)

a_train.shape,a_test.shape
```

```python
#データのシャッフルをFalseで無効に。通常はシャッフルするほうが好ましい。
a_train, a_test = train_test_split(a,test_size=0.2, shuffle=False)

a_train.shape,a_test.shape
```

```python
a_train
```

```python
#コード実行時毎回ランダム生成されるので、指定する
#2つの配列を同時につくる→元のaのn項目と元のbのn項目の対応関係が維持できる
a_train, a_test, b_train, b_test = train_test_split(a,b, test_size=0.2,random_state=365)

a_train.shape,a_test.shape
```
